```python
SECRET_CATALOG = "capstone_8hospital"
SECRET_SCHEMA = "bronze_schema"
jdbc_user = dbutils.secrets.get(catalog=SECRET_CATALOG, schema=SECRET_SCHEMA, key="sql-user")
jdbc_password = dbutils.secrets.get(catalog=SECRET_CATALOG, schema=SECRET_SCHEMA, key="sql-pwd")

jdbc_url=(
    "jdbc:sqlserver://capstone8dbserver.database.windows.net:1433;"
    "database=capstone8;"
    "encrypt=true;"
    "trustServerCertificate=false;"
    "hostNameInCertificate=*.database.windows.net;"
    "loginTimeout=30;"
    )
```

```python
patient_df = (spark.read
    .format("jdbc")
    .option("url", jdbc_url)
    .option("dbtable", "dbo.patient_master")
    .option("user", jdbc_user)
    .option("password", jdbc_password)
    .option("driver", "com.microsoft.sqlserver.jdbc.SQLServerDriver")
    .load())
display(patient_df)
```

```python
department_df = (spark.read
    .format("jdbc")
    .option("url", jdbc_url)
    .option("dbtable", "dbo.department_master")
    .option("user", jdbc_user)
    .option("password", jdbc_password)
    .option("driver", "com.microsoft.sqlserver.jdbc.SQLServerDriver")
    .load())
display(department_df)
```

```python
bed_df = (spark.read
    .format("jdbc")
    .option("url", jdbc_url)
    .option("dbtable", "dbo.bed_master")
    .option("user", jdbc_user)
    .option("password", jdbc_password)
    .option("driver", "com.microsoft.sqlserver.jdbc.SQLServerDriver")
    .load())
display(bed_df)
```

```python
catalog = "capstone_8hospital"
schema = "bronze_schema"

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")

patient_df.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.bronze_patients")
department_df.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.bronze_departments")
bed_df.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.bronze_beds")
```
